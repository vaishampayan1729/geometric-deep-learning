## A.1. Manifolds and Smooth Maps

An $n$-dimensional manifold is a Hausdorff, second-countable topological space locally homeomorphic to open subsets of $\mathbb R^n$. A **chart** is a pair $(U,\varphi)$ with $U\subset M$ open and $\varphi:U\to\varphi(U)\subset\mathbb R^n$ a homeomorphism. An atlas covers $M$ by charts. Charts are smoothly compatible when their transition maps are smooth; a smooth structure is a maximal compatible atlas.

A function $f:M\to\mathbb R$ is smooth when $f\circ\varphi^{-1}$ is smooth in every chart. A map $F:N\to M$ is smooth when its coordinate representations $\psi\circ F\circ\varphi^{-1}$ are smooth. A **diffeomorphism** is a smooth bijection with smooth inverse.

**Example: the circle.** $S^1=\{(x,y):x^2+y^2=1\}$ is a 1-manifold, despite being embedded in $\mathbb R^2$. On the arc $x>0$, the coordinate $y$ determines $x=\sqrt{1-y^2}$. Several such charts cover the circle; their transition maps are smooth on overlaps.

Manifolds are locally Euclidean, but their global topology can be nontrivial. Since manifolds are locally path-connected, their connected components are open; hence a connected manifold is path-connected. This implication is not true for arbitrary topological spaces.

## A.2. Tangent Vectors

A tangent vector at $p$ can be defined intrinsically as a **derivation** on germs of smooth functions. If $C_p^\infty(M)$ denotes the algebra of smooth-function germs at $p$, a tangent vector is a linear map $v:C_p^\infty(M)\to\mathbb R$ satisfying

$$v(fg)=v(f)g(p)+f(p)v(g).$$

The tangent space $T_pM$ is the vector space of such derivations. The product rule captures the essential behavior of directional differentiation.

For a chart $(U,x^1,\ldots,x^n)$, the coordinate tangent vectors $\left.\partial/\partial x^i\right|_p$ form a basis, and

$$v=\sum_i v^i\left.\frac{\partial}{\partial x^i}\right|_p,\qquad v(f)=\sum_i v^i\frac{\partial f}{\partial x^i}(p).$$

A smooth curve $\gamma$ with $\gamma(0)=p$ defines $\dot\gamma(0)(f)=\left.\frac{d}{dt}\right|_0 f(\gamma(t))$.

For a smooth map $F:N\to M$, its differential is the linear map $dF_p:T_pN\to T_{F(p)}M$ defined by

$$(dF_pv)(f)=v(f\circ F).$$

It generalizes the Jacobian matrix and obeys the chain rule $d(G\circ F)_p=dG_{F(p)}\circ dF_p$.

## A.3. Vector Fields

A smooth vector field assigns a tangent vector $X_p\in T_pM$ to each point, smoothly in local coordinates. In a chart,

$$X=\sum_i a^i\frac{\partial}{\partial x^i},$$

where the coefficient functions $a^i$ are smooth. It acts on functions as a first-order differential operator:

$$Xf=\sum_i a^i\frac{\partial f}{\partial x^i},\qquad X(fg)=(Xf)g+f(Xg).$$

Vector fields form a module over $C^\infty(M)$. Their commutator defines the **Lie bracket**

$$[X,Y]f=X(Yf)-Y(Xf).$$

If $X=\sum_i a^i\partial_i$ and $Y=\sum_i b^i\partial_i$, then

$$[X,Y]=\sum_j\left(\sum_i a^i\partial_i b^j-b^i\partial_i a^j\right)\partial_j.$$

The bracket is bilinear, antisymmetric, and satisfies the Jacobi identity. It measures the failure of the differential operators to commute. A vector field also determines local integral curves satisfying $\dot\gamma(t)=X_{\gamma(t)}$.

**Example.** On $\mathbb R^2$, for $X=\partial_x$ and $Y=x\partial_y$, one has $[X,Y]=\partial_y$.

## A.4. Differential Forms

The cotangent space is $T_p^*M=(T_pM)^*$. A **differential $k$-form** assigns to each point a smoothly varying alternating $k$-linear map $(T_pM)^k\to\mathbb R$. A 0-form is a smooth function, a 1-form is a covector field, and a 2-form evaluates oriented pairs of tangent vectors.

For coordinates $(x^1,\ldots,x^n)$, the coordinate 1-forms are dual to coordinate tangent vectors:

$$dx^i\left(\frac{\partial}{\partial x^j}\right)=\delta^i_j.$$

Every $k$-form has a local expansion

$$\omega=\sum_{i_1<\cdots<i_k}a_{i_1\cdots i_k}\,dx^{i_1}\wedge\cdots\wedge dx^{i_k},$$

with smooth coefficients. The wedge product is alternating: $dx^i\wedge dx^j=-dx^j\wedge dx^i$ and $dx^i\wedge dx^i=0$.

**Examples.** On $\mathbb R^2$, $\alpha=P\,dx+Q\,dy$ is a 1-form. On $\mathbb R^3$, $\omega=P\,dy\wedge dz+Q\,dz\wedge dx+R\,dx\wedge dy$ is a 2-form. Forms evaluate tangent directions and encode oriented infinitesimal quantities.

## A.5. Exterior Differentiation on a Manifold

The exterior derivative maps $k$-forms to $(k+1)$-forms:

$$d:\Omega^k(M)\longrightarrow\Omega^{k+1}(M).$$

For a function $f$,

$$df=\sum_i\frac{\partial f}{\partial x^i}\,dx^i.$$

If $\omega=\sum_I a_I\,dx^{i_1}\wedge\cdots\wedge dx^{i_k}$, then

$$d\omega=\sum_I da_I\wedge dx^{i_1}\wedge\cdots\wedge dx^{i_k}.$$

It obeys the graded Leibniz rule and nilpotence:

$$d(\omega\wedge\eta)=d\omega\wedge\eta+(-1)^k\omega\wedge d\eta,\qquad d^2=0,$$

where $\omega$ has degree $k$. A form is **closed** if $d\omega=0$ and **exact** if $\omega=d\eta$ for some $\eta$. Every exact form is closed, though a closed form need not be exact globally.

**Coordinate example.** For $\alpha=P\,dx+Q\,dy$,

$$d\alpha=\left(\frac{\partial Q}{\partial x}-\frac{\partial P}{\partial y}\right)dx\wedge dy.$$

Although expressed locally in coordinates, $d\omega$ is intrinsically defined and does not depend on the chart.

## A.6. Exterior Differentiation on $\mathbb R^3$

Use the Euclidean metric to identify a vector field $\mathbf F=(P,Q,R)$ with its dual 1-form

$$\mathbf F^\flat=P\,dx+Q\,dy+R\,dz.$$

**Gradient.** For a scalar function $f$,

$$df=f_x\,dx+f_y\,dy+f_z\,dz,$$

which corresponds to $\nabla f=(f_x,f_y,f_z)$ under the metric identification.

**Curl.** Differentiating the dual 1-form gives

$$\begin{aligned}d\mathbf F^\flat={}&(Q_x-P_y)\,dx\wedge dy\\&+(R_x-P_z)\,dx\wedge dz\\&+(R_y-Q_z)\,dy\wedge dz.\end{aligned}$$

The Hodge star, which depends on the Euclidean metric and orientation, converts this 2-form to the 1-form corresponding to $\nabla\times\mathbf F=(R_y-Q_z,P_z-R_x,Q_x-P_y)$.

**Divergence.** The flux 2-form is

$$\iota_{\mathbf F}(dx\wedge dy\wedge dz)=P\,dy\wedge dz+Q\,dz\wedge dx+R\,dx\wedge dy.$$

Its exterior derivative is

$$d\bigl(\iota_{\mathbf F}\mathrm{vol}\bigr)=(P_x+Q_y+R_z)\,dx\wedge dy\wedge dz,$$

so the coefficient is $\nabla\cdot\mathbf F$. Thus $d$ unifies vector-calculus operations; the metric and orientation enter when translating between forms and vectors.

## A.7. Pullback of Differential Forms

Let $F:N\to M$ be smooth. The pullback transports forms from $M$ to $N$. For a function, $F^*f=f\circ F$. For a $k$-form $\omega$,

$$(F^*\omega)_p(v_1,\ldots,v_k)=\omega_{F(p)}(dF_pv_1,\ldots,dF_pv_k).$$

The tangent vectors are first mapped by $dF_p$, then evaluated by $\omega$. Pullback is always defined, even if $F$ is not invertible. In general, vector fields do not admit a corresponding pushforward along arbitrary smooth maps.

Pullback respects wedge products and commutes with exterior differentiation:

$$F^*(\omega\wedge\eta)=F^*\omega\wedge F^*\eta,\qquad d(F^*\omega)=F^*(d\omega).$$

In coordinates, $F^*(dx^i)=d(x^i\circ F)$. For a curve $F(t)=(x(t),y(t))$ and a 1-form $\omega=P\,dx+Q\,dy$,

$$F^*\omega=\bigl(P(x(t),y(t))x'(t)+Q(x(t),y(t))y'(t)\bigr)dt.$$

This is the usual integrand for a line integral. If $i:S\hookrightarrow M$ is an inclusion, $i^*\omega$ is the restriction of the form to tangent directions along $S$, often denoted $\omega|_S$.

## Summary: the structure of the appendix

- **Manifolds and smooth maps:** local Euclidean coordinates with smooth transitions.
- **Tangent vectors:** derivations on smooth-function germs; $dF_p$ is the intrinsic Jacobian.
- **Vector fields:** smooth sections of the tangent bundle and derivations of functions; their commutator is the Lie bracket.
- **Differential forms:** alternating covariant tensors with wedge products.
- **Exterior derivative:** $d^2=0$ and the graded Leibniz rule.
- **Three-dimensional vector calculus:** gradient, curl, and divergence emerge from $d$ plus metric/orientation identifications.
- **Pullback:** forms transform contravariantly, preserving wedge products and commuting with $d$.

These constructions are prerequisites for expressing connections and curvature using differential forms.
